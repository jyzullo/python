```python
def f_to_c(degrees_f):
    #convert degrees Fahrenheit to degrees Celsius
    degrees_f = (degrees_f - 32) * 5 / 9
    return degrees_f

f_to_c(32)  
f_to_c(100)
```

```python
print(f_to_c(32))
```

```python
def double_it(multiply_by_2):
    # doubles the number given to it    
    result = multiply_by_2 * 2
    return result
```

```python
print(double_it(43))
```
